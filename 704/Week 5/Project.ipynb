## Part 1: Searching vs Games

Consider the state space illustrated below.
Each terminal state state is marked with a reward for reaching that state.
Each non-terminal state has two possible actions represented by the two outgoing arrows to later (lower) states.
The only rewards are for reaching the terminal states, there are no diminishing returns (i.e. $\gamma=1$), and there is no randomness so actions may be freely chosen.

![](https://github.com/bu-cds-dx704/dx704-project-05/blob/main/part1.png?raw=true)

Solve for the value of each non-terminal state according to the following three scenarios.

1. Search: There is one agent that picks all actions with the goal of maximizing the final reward.
2. Minimax: There are two agents P1 and P2. P1 controls the actions for the 1st and 3rd rows (i.e. the states marked A and D-G) while P2 controls the actions for the 2nd and 4th rows (i.e. the states B-C and H-J). P1 seeks to maximize the final reward, and P2 seeks to minimize the final reward.
3. Maximin: P1 and P2 control the same states as before, but P1 seeks to minimize the final reward, and P2 seeks to maximize the final reward.

Save your results in a file "values-1.tsv" with the column state with label A-J and columns search_value, minimax_value, and maximin_value that respectively correspond to the three scenarios.

Hint: Print out the image above and compute the values by hand in a bottom up fashion.

Submit the file "values-1.tsv" in Gradescope.

```python
# Part 1: Searching vs Games - Game Tree Solution

# Define the game tree structure
# Each state maps to [left_child, right_child]
game_tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J', '+5'],
    'F': ['+1', '-1'],
    'G': ['+1', '0'],
    'H': ['+9', '-1'],
    'I': ['+20', '+1'],
    'J': ['+3', '+2']
}

# P1 controls rows 1 and 3: A, D, E, F, G
# P2 controls rows 2 and 4: B, C, H, I, J
p1_states = {'A', 'D', 'E', 'F', 'G'}
p2_states = {'B', 'C', 'H', 'I', 'J'}

def is_terminal(state):
    """Check if state is terminal (not in the main state list)"""
    return state not in game_tree

def get_terminal_value(state):
    """Get the value of a terminal state"""
    return float(state)

# Scenario 1: SEARCH (Single Agent Maximizing)
def compute_search_value(state, memo):
    """Compute value for search scenario - single agent maximizing"""
    if state in memo:
        return memo[state]
    
    if is_terminal(state):
        value = get_terminal_value(state)
    else:
        children = game_tree[state]
        left_val = compute_search_value(children[0], memo)
        right_val = compute_search_value(children[1], memo)
        value = max(left_val, right_val)  # Always maximize
    
    memo[state] = value
    return value

# Scenario 2: MINIMAX (P1 Maximizes, P2 Minimizes)
def compute_minimax_value(state, memo):
    """Compute value for minimax scenario - P1 max, P2 min"""
    if state in memo:
        return memo[state]
    
    if is_terminal(state):
        value = get_terminal_value(state)
    else:
        children = game_tree[state]
        left_val = compute_minimax_value(children[0], memo)
        right_val = compute_minimax_value(children[1], memo)
        
        if state in p1_states:
            value = max(left_val, right_val)  # P1 maximizes
        else:
            value = min(left_val, right_val)  # P2 minimizes
    
    memo[state] = value
    return value

# Scenario 3: MAXIMIN (P1 Minimizes, P2 Maximizes)
def compute_maximin_value(state, memo):
    """Compute value for maximin scenario - P1 min, P2 max"""
    if state in memo:
        return memo[state]
    
    if is_terminal(state):
        value = get_terminal_value(state)
    else:
        children = game_tree[state]
        left_val = compute_maximin_value(children[0], memo)
        right_val = compute_maximin_value(children[1], memo)
        
        if state in p1_states:
            value = min(left_val, right_val)  # P1 minimizes
        else:
            value = max(left_val, right_val)  # P2 maximizes
    
    memo[state] = value
    return value

# Compute all values for all states
search_memo = {}
minimax_memo = {}
maximin_memo = {}

search_values = {}
minimax_values = {}
maximin_values = {}

for state in ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']:
    search_values[state] = compute_search_value(state, search_memo)
    minimax_values[state] = compute_minimax_value(state, minimax_memo)
    maximin_values[state] = compute_maximin_value(state, maximin_memo)

# Display results
print("=" * 60)
print("SCENARIO 1: SEARCH (Single Agent Maximizing)")
print("=" * 60)
for state in ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']:
    print(f"  {state}: {int(search_values[state])}")

print("\n" + "=" * 60)
print("SCENARIO 2: MINIMAX (P1 Maximizes, P2 Minimizes)")
print("=" * 60)
for state in ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']:
    print(f"  {state}: {int(minimax_values[state])}")

print("\n" + "=" * 60)
print("SCENARIO 3: MAXIMIN (P1 Minimizes, P2 Maximizes)")
print("=" * 60)
for state in ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']:
    print(f"  {state}: {int(maximin_values[state])}")

# Save results to TSV file
print("\n" + "=" * 60)
print("CREATING OUTPUT FILE")
print("=" * 60)

output_path = 'values-1.tsv'
with open(output_path, 'w') as f:
    f.write('state\tsearch_value\tminimax_value\tmaximin_value\n')
    for state in ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']:
        search_val = int(search_values[state])
        minimax_val = int(minimax_values[state])
        maximin_val = int(maximin_values[state])
        f.write(f'{state}\t{search_val}\t{minimax_val}\t{maximin_val}\n')

print(f"\nÃƒÂ¢Ã…â€œÃ¢â‚¬Å“ Output saved to '{output_path}'")

# Display the TSV content
print("\nResults:")
with open(output_path, 'r') as f:
    print(f.read())
```

## Part 2: Picking Up Sticks

The state space illustrated below corresponds to a variation of the game [Nim](https://en.wikipedia.org/wiki/Nim).
States labeled with a prefix of "p1_" correspond to states where player P1 chooses the action while states labeled with a prefix of "p2_" correspond to states where player P2 chooses the action.
The number in the suffix is the number of "sticks" remaining.
The players take turns choosing actions, and each action corresponds to removing one or two sticks.
When there are no more sticks, the player who would have picked an action loses.


![](https://github.com/bu-cds-dx704/dx704-project-05/blob/main/part2.png?raw=true)

For example, from the state labeled "p1_1", there is one stick left, player P1 removes the last stick, and player P2 loses.
The loss for P2 is represented by a final reward of +1.
A loss for P1 is represented by a final reward of -1.
Player P1 tries to maximize the final reward, and player P2 tries to minimize the final reward.

Solve for the value of each of the non-terminal states.
Save the results in a file "values-2.tsv" with columns state and value.

## Part 3: Searching a Maze

Consider the following maze.

![](https://github.com/bu-cds-dx704/dx704-project-05/blob/main/part3.png?raw=true)

State C is a terminal state giving reward +100.
The remaining states have a reward of -1 when they are reached.
So moving to state F has a value of +99 due to the reward of -1 at state F and the optimal action of moving to state C for the reward of +100 afterwards.

Compute the values for states A-J and S and save them in a file "values-3.tsv" with columns state and value.

```python
# Part 3: Searching a Maze

# Define the maze graph from the diagram
maze_graph = {
    'A': ['S'],           # A -> S
    'S': ['A', 'B'],      # S <-> A, S <-> B
    'B': ['S', 'D'],      # B <-> S, B <-> D
    'D': ['B', 'E', 'G'], # D <-> B, D <-> E, D <-> G
    'E': ['D', 'H'], # E <-> D, E <-> H, E -> I
    'H': ['E', 'I'],      # H <-> E, H -> I
    'G': ['D', 'I'],      # G <-> D, G -> I
    'I': ['G', 'H', 'J'], # I from G,E,H and I <-> J
    'J': ['I', 'F'],      # J <-> I, J <-> F
    'F': ['J', 'C'],      # F <-> J, F -> C
    'C': []               # Terminal state
}

def get_reward(state):
    if state == 'C':
        return 100.0
    else:
        return -1.0

def compute_maze_values(graph, max_iterations=100, tolerance=1e-9, discount=0.99):
    states = list(graph.keys())
    values = {state: 0.0 for state in states}
    values['C'] = 100.0
    
    for iteration in range(max_iterations):
        old_values = values.copy()
        
        for state in states:
            if state == 'C':
                values[state] = 100.0
            else:
                next_states = graph[state]
                if next_states:
                    best_next_value = max(values[next_state] for next_state in next_states)
                    values[state] = -1.0 + (discount * best_next_value)
                else:
                    values[state] = -1.0
        
        max_diff = max(abs(values[s] - old_values[s]) for s in states)
        if max_diff < tolerance:
            break
    
    return values

maze_values = compute_maze_values(maze_graph)

print('=' * 60)
print('MAZE SEARCH - State Values')
print('=' * 60)
for state in ['A', 'S', 'B', 'D', 'E', 'H', 'G', 'I', 'J', 'F', 'C']:
    if state in maze_values:
        print(f'  {state}: {maze_values[state]}')

print('\nMaze Graph Structure:')
for state in sorted(maze_graph.keys()):
    print(f'  {state} -> {sorted(maze_graph[state])}')
```

```python
print('\n' + '=' * 60)
print('CREATING OUTPUT FILE')
print('=' * 60)

output_path_3 = 'values-3.tsv'
with open(output_path_3, 'w') as f:
    f.write('state\tvalue\n')
    for state in ['A', 'S', 'B', 'D', 'E', 'H', 'G', 'I', 'J', 'F', 'C']:
        f.write(f'{state}\t{maze_values[state]}\n')

print(f"\nOutput saved to '{output_path_3}'")
print('\nResults:')
with open(output_path_3, 'r') as f:
    print(f.read())
```
