#### Utility Function to Pretty-Print Sentences

````python
'''
Reconstruct a human-readable sentence from tokens.
- Capitalize the first token (after removing <s>).
- Attach any token in string.punctuation with no preceding space.
- Attach tokens starting with an apostrophe (e.g., "'s", "'ll") with no space.
- Capitalize standalone 'i' and 'god' wherever they appear.
- If the last token is '.', '?', or '!', attach it without a space.
- Ignore boundary markers <s> and </s> if present.
Returns the sentence string and also prints it.
'''

def print_sentence(tokens):

    if not tokens:
        print("")
        return ""

    # Drop boundary markers if present
    start = 1 if tokens and tokens[0] == "<s>" else 0
    end = -1 if tokens and tokens[-1] == "</s>" else len(tokens)
    core = tokens[start:end]

    if not core:
        print("")
        return ""

    # Capitalize first token
    out = core[0].capitalize()

    # Process remaining tokens with spacing rules
    for i, t in enumerate(core[1:], start=1):
        is_last = (i == len(core) - 1)

        # Last-token punctuation rule
        if is_last and t in {".", "?", "!"}:
            out += t
            continue

        # General punctuation attaches with no space (e.g., ',', ':', ';', ')', etc.)
        if t in string.punctuation:
            out += t
            continue

        # Apostrophe-start tokens like "'s", "'d", "'ll"
        if t.startswith("'"):
            out += t
            continue

        # Capitalize special standalone words
        if t.lower() in {"i", "god"}:
            out += " " + t.capitalize()
            continue

        # Default: add a space before normal words
        out += " " + t

    print(out)
    return out
````

### Digression: Zipf's Law

Zipf's Law is an empirical regularity of natural language:  
if you rank words by their frequency in a corpus, the frequency \( f(r) \) of the word at rank \( r \) is approximately inversely proportional to its rank:

$$
f(r) \propto \frac{1}{r^k}
$$

where $ r = 1, 2, 3, \dots $ is the rank of a word (1 = most frequent), and $ k \approx 1 $.  

- The most common words (e.g., *the*, *of*, *and*) occur very often.  
- A handful of mid-ranked words occur moderately often.  
- The vast majority of words occur rarely.  

On a **log–log plot** of frequency vs. rank, this relationship appears as a straight line with slope $-k$.  
For natural language, empirical studies show that $k \approx 1$, which is why we plot below a reference line with slope $-1$ to compare against real corpus data.


**Why this matters:** Because most words are rare, language models must use techniques like **smoothing, subword modeling, or embeddings** to handle sparsity and represent meaning effectively.  


````python
from collections import Counter

# --------------------------------------------------
# Word frequencies
# --------------------------------------------------

# Count words only, excluding punctuation and
# sentence-boundary markers
word_counts = Counter(
    token
    for sentence in tokenized_sentences
    for token in sentence
    if is_word(token)
)

# Frequencies sorted by rank
freqs = np.array([
    freq
    for token, freq in word_counts.most_common()
])

ranks = np.arange(1, len(freqs) + 1)

plt.figure(figsize=(8, 5))
plt.plot(ranks, freqs, label="Corpus data")

# Reference line: slope = -1 (Zipf's law)
# Match scale by anchoring at the first frequency
ref_line = freqs[0] / ranks
plt.plot(
    ranks,
    ref_line,
    "--",
    label="Reference slope = -1"
)

plt.xscale("log")
plt.yscale("log")
plt.title("Word Frequencies vs. Rank (Zipf's Law)")
plt.xlabel("Rank of word")
plt.ylabel("Frequency")
plt.grid(True, which="both", ls="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()


# --------------------------------------------------
# Sentence lengths
# --------------------------------------------------

# Subtract the <s> and </s> boundary tokens
sentence_lengths = [
    len(sentence) - 2
    for sentence in tokenized_sentences
]

plt.figure(figsize=(8, 4))
plt.hist(
    sentence_lengths,
    bins=range(1, max(sentence_lengths) + 2),
    color="skyblue",
    edgecolor="black"
)

plt.title("Histogram of Sentence Lengths")
plt.xlabel("Sentence length (in tokens)")
plt.ylabel("Number of sentences")
plt.tight_layout()
plt.show()

print(
    f"Average sentence length: "
    f"{np.mean(sentence_lengths):.2f} tokens"
)

print(
    f"Max sentence length:     "
    f"{max(sentence_lengths)} tokens"
)
````

````python
# Your code here (add as many cells as you need)

# PART A: Build N-grams for N = 2 and N = 3


from collections import defaultdict
import math
import random

def build_ngrams(tokenized_sentences, n):
    """
    Build n-grams from a list of tokenized sentences.
    Each sentence should be enclosed by <s> and </s> tokens.
    """
    ngrams = []
    for sentence in tokenized_sentences:
        for i in range(len(sentence) - n + 1):
            ngram = tuple(sentence[i:i+n])
            ngrams.append(ngram)
    return ngrams

# Build bigrams (N=2) and trigrams (N=3)
bigrams = build_ngrams(tokenized_sentences, 2)
trigrams = build_ngrams(tokenized_sentences, 3)

print(f"Number of bigrams: {len(bigrams)}")
print(f"Number of trigrams: {len(trigrams)}")
print(f"First 5 bigrams: {bigrams[:5]}")
print(f"First 5 trigrams: {trigrams[:5]}")

# PART B: Create dictionary mapping left contexts to next-token lists


next_word_lists = defaultdict(list)

# Process bigrams (N=2): left context is 1 token, next token is 1 token
for bigram in bigrams:
    left_context = bigram[:-1]  # first N-1 = 1 token
    next_token = bigram[-1]      # last token
    next_word_lists[left_context].append(next_token)

# Process trigrams (N=3): left context is 2 tokens, next token is 1 token
for trigram in trigrams:
    left_context = trigram[:-1]  # first N-1 = 2 tokens
    next_token = trigram[-1]      # last token
    next_word_lists[left_context].append(next_token)

# Convert defaultdict to regular dict for clarity
next_word_lists = dict(next_word_lists)

print(f"\nNumber of unique contexts: {len(next_word_lists)}")
print(f"Example contexts and their next tokens:")
for i, (context, tokens) in enumerate(list(next_word_lists.items())[:3]):
    print(f"  {context}: {tokens}")

# PART C: Convert to log-probability distributions

def compute_log_probabilities(next_tokens_list):
    """
    Convert a list of next tokens to a log-probability distribution.
    """
    # Count occurrences
    token_counts = defaultdict(int)
    for token in next_tokens_list:
        token_counts[token] += 1
    
    total = len(next_tokens_list)
    log_probs = {}
    
    for token, count in token_counts.items():
        prob = count / total
        log_prob = math.log(prob)
        log_probs[token] = log_prob
    
    return log_probs

# PART D: Create the Probability Language Model

get_next_word = {}

for context, next_tokens_list in next_word_lists.items():
    get_next_word[context] = compute_log_probabilities(next_tokens_list)

print(f"\nProbability Language Model created!")
print(f"Number of contexts in PLM: {len(get_next_word)}")

# PART E: Sanity-check your model

contexts_to_check = [('<s>',), ('the',), ('<s>', 'he')]

print("\n" + "="*60)
print("PART E: Sanity-Checking Model")
print("="*60)

for context in contexts_to_check:
    if context in get_next_word:
        print(f"\nContext: {context}")
        print(f"Log-probability distribution:")
        for next_token, log_prob in sorted(
            get_next_word[context].items(), 
            key=lambda x: x[1], 
            reverse=True
        ):
            print(f"  {next_token:15} : log_prob = {log_prob:8.4f}")
    else:
        print(f"\nContext {context} not found in model.")
        print(f"Showing a similar context instead...")
        # Find a context with the same length
        for ctx in get_next_word.keys():
            if len(ctx) == len(context):
                print(f"Context: {ctx}")
                print(f"Log-probability distribution:")
                for next_token, log_prob in sorted(
                    get_next_word[ctx].items(), 
                    key=lambda x: x[1], 
                    reverse=True
                )[:5]:  # Show top 5
                    print(f"  {next_token:15} : log_prob = {log_prob:8.4f}")
                break

# PART F: Sample possible next tokens

def sample_next_tokens(context, model, num_samples=5, random_seed=42):
    """
    Sample possible next tokens from the model for a given context.
    Converts log probabilities back to ordinary probabilities.
    """
    if context not in model:
        return None
    
    rng = random.Random(random_seed)
    log_prob_dict = model[context]
    
    # Convert log probabilities back to ordinary probabilities
    tokens = list(log_prob_dict.keys())
    log_probs = [log_prob_dict[token] for token in tokens]
    
    # Convert from log space: prob = exp(log_prob)
    probs = [math.exp(lp) for lp in log_probs]
    
    # Sample from the distribution
    samples = rng.choices(tokens, weights=probs, k=num_samples)
    
    return samples

print("\n" + "="*60)
print("PART F: Sampling Next Tokens")
print("="*60)

for context in contexts_to_check:
    if context in get_next_word:
        samples = sample_next_tokens(context, get_next_word, num_samples=5)
        print(f"\nContext: {context}")
        print(f"5 sampled next tokens: {samples}")
    else:
        # Find a similar context to sample from
        for ctx in get_next_word.keys():
            if len(ctx) == len(context):
                samples = sample_next_tokens(ctx, get_next_word, num_samples=5)
                print(f"\nContext: {ctx} (instead of {context})")
                print(f"5 sampled next tokens: {samples}")
                break
````

## Problem Two — Generating Sentences with Backoff and Temperature

In this problem, you will use the probabilistic language model from Problem One to generate complete sentences.

At each step, the generator examines the most recent tokens in the sentence, finds the corresponding conditional probability distribution, samples the next token, and continues until the end-of-sentence token `</s>` is generated.

For **bigrams**, this is straightforward: each token observed in the corpus has at least one possible continuation.

For **trigrams**, two additional issues must be handled.

At the beginning of a sentence, there are not yet two previous tokens available to form a trigram context.

For example:

```text

Step 2:    <s> one ?                 trigram
Step 3:    <s> one fish ?            trigram
```

Your code can handle this naturally by using whatever left context is currently available.

For example:


### 2. Sparsity: Missing Trigram Contexts

Even after two previous tokens are available, their trigram context may never have appeared in the corpus.

For example, suppose the sentence generated so far is:


If the context


In this case, use **backoff**:

> If the trigram context does not exist, drop the earliest token and use the corresponding bigram context instead.

Thus, if:



With the bigram and trigram models constructed in Problem One, the basic decision is:

### Temperature-Controlled Sampling

Once you have selected the appropriate probability distribution, use **temperature scaling** before sampling the next token.

Temperature controls how strongly the generator favors high-probability choices:

- **Low temperature** concentrates probability on the most likely tokens, producing more conservative and often more repetitive text.
- **Temperature = 1.0** uses the model probabilities without changing their relative shape.
- **High temperature** produces a flatter distribution, increasing variety but usually reducing coherence.

Your PLM stores probabilities in **log space**, so remember to convert them back to ordinary probabilities before applying temperature scaling and using them as sampling weights.

````python
# PART A: Temperature-Controlled Sampling Function

import math
import random

def sample_next_token(log_prob_dict, temperature=1.0, random_seed=None):
    """
    Sample the next token from a log-probability distribution.
    
    Args:
        log_prob_dict: Dictionary mapping tokens to log probabilities
        temperature: Temperature for scaling (1.0 = no scaling)
        random_seed: Optional seed for reproducibility
    
    Returns:
        A sampled token
    """
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    # Extract tokens and log probabilities
    tokens = list(log_prob_dict.keys())
    log_probs = [log_prob_dict[token] for token in tokens]
    
    # Convert from log space to ordinary probabilities
    probs = [math.exp(log_prob) for log_prob in log_probs]
    
    # Apply temperature scaling
    # log(p^(1/T)) = (1/T) * log(p)
    # So: exp((1/T) * log(p)) = p^(1/T)
    if temperature != 1.0:
        scaled_log_probs = [log_prob / temperature for log_prob in log_probs]
        scaled_probs = [math.exp(lp) for lp in scaled_log_probs]
        # Normalize to make sure they sum to 1
        total = sum(scaled_probs)
        scaled_probs = [p / total for p in scaled_probs]
    else:
        scaled_probs = probs
    
    # Sample from the distribution
    sampled_token = rng.choices(tokens, weights=scaled_probs, k=1)[0]
    
    return sampled_token


# PART B: Sentence Generation with Backoff

def generate_sentence(get_next_word, temperature=1.0, max_length=40, random_seed=None):
    """
    Generate a complete sentence using the PLM with trigram/bigram backoff.
    
    Args:
        get_next_word: The PLM dictionary from Problem One
        temperature: Temperature for sampling (1.0 = model probabilities)
        max_length: Maximum tokens to generate (safety limit)
        random_seed: Optional seed for reproducibility
    
    Returns:
        A list of tokens representing the generated sentence
    """
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    sentence = ['<s>']
    
    for step in range(max_length):
        # Get the left context from the last 2 tokens
        # This naturally gives us a 1-token context at the start
        # and a 2-token context thereafter
        left_context = tuple(sentence[-2:])
        
        # Try trigram context first
        if left_context in get_next_word:
            log_prob_dist = get_next_word[left_context]
        else:
            # Backoff to bigram context (drop earliest token)
            left_context = tuple(sentence[-1:])
            if left_context in get_next_word:
                log_prob_dist = get_next_word[left_context]
            else:
                # Should not happen with proper construction
                print(f"Warning: No context found for {left_context}")
                break
        
        # Sample the next token using temperature-controlled sampling
        next_token = sample_next_token(log_prob_dist, temperature, random_seed)
        sentence.append(next_token)
        
        # Stop if we generated the end-of-sentence marker
        if next_token == '</s>':
            break
    
    return sentence


# PART C: Generate and Compare Sentences

# Generate 10 sentences at each temperature
temperatures = [0.2, 1.0, 5.0]
num_sentences_per_temp = 10

results = {}

for temp in temperatures:
    print(f"\n{'='*60}")
    print(f"Temperature: {temp}")
    print(f"{'='*60}\n")
    
    results[temp] = []
    
    for i in range(num_sentences_per_temp):
        sentence = generate_sentence(get_next_word, temperature=temp, random_seed=42+i)
        results[temp].append(sentence)
        
        print(f"Sentence {i+1}:")
        print_sentence(sentence)
        print()


# Qualitative Analysis

print("\n" + "="*60)
print("QUALITATIVE COMPARISON")
print("="*60 + "\n")

print("""
Temperature 0.2 (Low - Conservative):
- Expected: High probability of repetition, conservative choices
- Likely to follow the most common patterns in the corpus
- May produce short, formulaic sentences
- Grammar usually correct but less variety

Temperature 1.0 (Model Probabilities):
- Expected: Balances naturalness with variety
- Follows the model's learned probabilities directly
- More diverse than 0.2 but still coherent
- Good baseline for comparison

Temperature 5.0 (High - Creative):
- Expected: Greater diversity, more experimental
- Flattens the probability distribution
- More likely to include unlikely word combinations
- May produce more grammatically awkward or incoherent sentences
- More "creative" or surprising word choices
""")

print("\nYour specific observations:")
print(f"- Temperature 0.2: Most repetitive? Most coherent? ")
print(f"- Temperature 1.0: Most balanced? Baseline quality? ")
print(f"- Temperature 5.0: Most diverse? Most unusual? ")
````

## Problem Three — Measuring Sentence Quality with Average Log Probability and Perplexity

The probabilities involved when we calculate the probability of longer token sequences can become extremely small, so numerical underflow is a concern. For this reason, we perform these calculations in **log space** (see **Appendix**).

At a basic level, we can evaluate a generated sentence by its **likelihood under the model**: sentences with higher probability are considered more likely by the model.

However, raw sentence probability is strongly affected by sentence length. Longer sentences involve more probability terms, so their total probability tends to become much smaller even when the individual word choices are reasonable.

To reduce this length effect, we use **Average Log Probability (ALP)** and **Perplexity**.

---

### Average Log Probability (ALP)

Let the tokenized sentence be

$$
[t_1, t_2, \dots, t_n]
$$

with

$$
t_1 = \texttt{<s>}
$$

and, for a completed sentence,

$$
t_n = \texttt{</s>}.
$$

For a trigram model with bigram backoff:

$$
\text{ALP}
=
\frac{
\log P(t_2 \mid t_1)
+
\log P(t_3 \mid t_1,t_2)
+
\cdots
+
\log P(t_n \mid t_{n-2},t_{n-1})
}{
n-1
}
$$

when the trigram contexts exist.

If a trigram context is unavailable, use the log probability from the corresponding **bigram** context instead.

### Notes

- We do not include the initial `<s>` token in the score because it is supplied to the model rather than predicted.
- A sequence containing \(n\) tokens including `<s>` contains exactly \(n-1\) predictions.
- ALP lies in

$$
(-\infty, 0]
$$

and **larger values are better**.
- A less negative ALP means that, on average, the model assigned higher probability to the generated tokens.

---

### Perplexity

Perplexity rescales ALP into a positive quantity:

```python
PP = np.exp(-ALP)
```

Perplexity is always at least 1:

$$
\text{PP} \ge 1
$$

and **smaller values are better**.

Intuitively, perplexity measures how uncertain or "confused" the model is when predicting the sequence. Lower perplexity means that the generated tokens were, on average, assigned higher probability by the model.

Because the exponential function is monotonic:

- maximizing ALP
- and minimizing perplexity

produce the same ranking of sentences.

> **Important:** In the remaining problems, use ALP to compare sequences of different lengths. When candidate sequences all have the same length, as they do within each step of beam search, accumulated log probability gives the same ranking. Use perplexity to report and compare completed sentences.

---

### Sampling and Scoring Are Different

Temperature changes the probability distribution used to **sample** the next token.

However, when evaluating the generated sentence, score each chosen token using the model's original **untempered log probability**.

In other words:

> **Sample using the temperature-adjusted probabilities, but score using the original model probabilities.**

This allows sentences generated at different temperatures to be evaluated using the same underlying language model.

---

### Carrying the Score During Generation

You can calculate the sentence score while the sentence is being generated.

Instead of storing only:

```python
tokens_so_far
```

maintain a pair:

```python
(tokens_so_far, sum_log_probs)
```

where `sum_log_probs` is the sum of the original log probabilities of all tokens generated so far.

For example:

```text
sum_log_probs =
    log P(t2 | t1)
  + log P(t3 | t1, t2)
  + ...
```

If backoff is used for a particular token, add the corresponding **bigram log probability** for that step.

The generation process should therefore follow this pattern:

1. Begin with:

```python
tokens = ['<s>']
sum_log_probs = 0.0
```

2. Determine the appropriate left context:
   - use the trigram context when available;
   - otherwise back off to the bigram context.

3. Sample the next token using the **temperature-adjusted** distribution.

4. Retrieve the sampled token's **original log probability** from `get_next_word`.

5. Add that value to `sum_log_probs`.

6. Append the token to the sentence.

7. Stop when `</s>` is generated, or when the **40-token maximum** from Problem Two is reached.

8. Compute:

```python
ALP = sum_log_probs / (len(tokens) - 1)
PP = np.exp(-ALP)
```

If generation stops because the 40-token limit is reached rather than because `</s>` was generated, compute ALP and perplexity over the tokens generated so far.

````python
# ================================================
# PART A: Modified Sampling Function
# ================================================

import math
import random
import numpy as np

def sample_next_token_with_score(log_prob_dict, temperature=1.0, random_seed=None):
    """
    Sample the next token and return both the token and its ORIGINAL log probability.
    
    Args:
        log_prob_dict: Dictionary mapping tokens to log probabilities
        temperature: Temperature for sampling (applied only for sampling, not scoring)
        random_seed: Optional seed for reproducibility
    
    Returns:
        (next_token, original_log_prob): The sampled token and its untempered log probability
    """
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    # Extract tokens and log probabilities
    tokens = list(log_prob_dict.keys())
    log_probs = [log_prob_dict[token] for token in tokens]
    
    # Convert from log space to ordinary probabilities
    probs = [math.exp(log_prob) for log_prob in log_probs]
    
    # Apply temperature scaling ONLY for sampling
    if temperature != 1.0:
        scaled_log_probs = [log_prob / temperature for log_prob in log_probs]
        scaled_probs = [math.exp(lp) for lp in scaled_log_probs]
        # Normalize to make sure they sum to 1
        total = sum(scaled_probs)
        scaled_probs = [p / total for p in scaled_probs]
    else:
        scaled_probs = probs
    
    # Sample from the temperature-adjusted distribution
    sampled_token = rng.choices(tokens, weights=scaled_probs, k=1)[0]
    
    # Get the ORIGINAL (untempered) log probability of the sampled token
    original_log_prob = log_prob_dict[sampled_token]
    
    return sampled_token, original_log_prob


# ================================================
# PART B: Modified Sentence Generator with Scoring
# ================================================

def generate_sentence_with_perplexity(get_next_word, temperature=1.0, max_length=40, random_seed=None):
    """
    Generate a complete sentence and compute its perplexity.
    
    Args:
        get_next_word: The PLM dictionary from Problem One
        temperature: Temperature for sampling (1.0 = model probabilities)
        max_length: Maximum tokens to generate (safety limit)
        random_seed: Optional seed for reproducibility
    
    Returns:
        (tokens, perplexity): The generated sentence and its perplexity score
    """
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    tokens = ['<s>']
    sum_log_probs = 0.0
    
    for step in range(max_length):
        # Get the left context from the last 2 tokens
        left_context = tuple(tokens[-2:])
        
        # Try trigram context first
        if left_context in get_next_word:
            log_prob_dist = get_next_word[left_context]
        else:
            # Backoff to bigram context (drop earliest token)
            left_context = tuple(tokens[-1:])
            if left_context in get_next_word:
                log_prob_dist = get_next_word[left_context]
            else:
                # Should not happen with proper construction
                print(f"Warning: No context found for {left_context}")
                break
        
        # Sample the next token and get its ORIGINAL log probability
        next_token, original_log_prob = sample_next_token_with_score(
            log_prob_dist, 
            temperature=temperature, 
            random_seed=random_seed
        )
        
        # Add the ORIGINAL log probability to our running score
        sum_log_probs += original_log_prob
        
        # Append token to sentence
        tokens.append(next_token)
        
        # Stop if we generated the end-of-sentence marker
        if next_token == '</s>':
            break
    
    # Compute ALP and Perplexity
    # Note: We have n tokens including <s>, so n-1 predictions
    num_predictions = len(tokens) - 1
    ALP = sum_log_probs / num_predictions
    PP = np.exp(-ALP)
    
    return tokens, PP


# ================================================
# PART C: Compare Three Temperatures with Perplexity
# ================================================

temperatures = [0.2, 1.0, 5.0]
num_sentences_per_temp = 10

results_with_perplexity = {}

for temp in temperatures:
    print(f"\n{'='*60}")
    print(f"Temperature: {temp}")
    print(f"{'='*60}\n")
    
    results_with_perplexity[temp] = []
    perplexities = []
    
    for i in range(num_sentences_per_temp):
        tokens, pp = generate_sentence_with_perplexity(
            get_next_word, 
            temperature=temp, 
            random_seed=42+i
        )
        
        results_with_perplexity[temp].append((tokens, pp))
        perplexities.append(pp)
        
        print(f"Sentence {i+1}: (Perplexity: {pp:.4f})")
        print_sentence(tokens)
        print()
    
    # Summary statistics for this temperature
    avg_pp = np.mean(perplexities)
    min_pp = np.min(perplexities)
    max_pp = np.max(perplexities)
    
    print(f"\nSummary for Temperature {temp}:")
    print(f"  Average Perplexity: {avg_pp:.4f}")
    print(f"  Min Perplexity:     {min_pp:.4f}")
    print(f"  Max Perplexity:     {max_pp:.4f}")
    print()


# ================================================
# Comparative Analysis
# ================================================

print("\n" + "="*70)
print("COMPARATIVE ANALYSIS: TEMPERATURE vs. PERPLEXITY")
print("="*70 + "\n")

print("""
KEY INSIGHT: Temperature affects HOW we sample, but perplexity measures 
how probable the RESULT is under the ORIGINAL model.

Expected Observations:

Temperature 0.2 (Low - Conservative):
- Favors high-probability tokens
- Sentences are repetitive, follow safe patterns
- PERPLEXITY: Should be LOWER (more probable under the model)
- Why? We're always picking the high-probability choices

Temperature 1.0 (Baseline):
- Uses unmodified model probabilities
- Balanced sentences with good variety
- PERPLEXITY: Moderate (baseline)
- This is the "true" model performance

Temperature 5.0 (High - Creative):
- Flattens distribution, includes low-probability tokens
- More diverse, sometimes incoherent
- PERPLEXITY: Should be HIGHER (less probable under the model)
- Why? We're including more unusual word combinations

COUNTERINTUITIVE FINDING:
The temperature that produces the most REALISTIC text (1.0) may NOT have 
the lowest perplexity! Lower temperatures can have lower perplexity but 
produce boring, repetitive text. Higher temperatures can have higher 
perplexity but seem more creative.

This demonstrates that perplexity measures LIKELIHOOD under the model, 
not HUMAN judgment of quality or realism.
""")

# Compare average perplexities across temperatures
print("\nAverage Perplexities by Temperature:")
for temp in temperatures:
    perplexities = [pp for tokens, pp in results_with_perplexity[temp]]
    avg_pp = np.mean(perplexities)
    print(f"  Temperature {temp}: {avg_pp:.4f}")
````

## Problem Four — Beam Search

For our final problem, you will implement **beam search**, a common strategy for finding high-probability sequences under a language model.

Beam search is a practical compromise between:

- **greedy decoding**, which keeps only the single best continuation at each step, and
- **exhaustive search**, which considers every possible sequence.

Instead, beam search keeps a small number of promising partial sentences and repeatedly expands them.



### The Basic Idea

Maintain a list of at most **B** incomplete sentences. This list is called the **beam** or priority queue `PQ`.

Each partial sentence should store:

```python
(tokens_so_far, sum_log_probs)
```

where `sum_log_probs` is the sum of the original, untempered log probabilities of the tokens generated so far.

Start with:

```python
PQ = [
    (['<s>'], 0.0)
]
```

At each iteration:

1. For every partial sentence currently in `PQ`, generate **C** possible one-token continuations.
2. If a continuation ends with `</s>`:
   - compute its perplexity;
   - move it to a separate list called `Finished`.
3. Otherwise:
   - keep it as an unfinished candidate.
4. Rank the unfinished candidates by their accumulated log probability.
5. Keep only the best **B** candidates and use them as the new `PQ`.

Repeat until you have collected 10 completed sentences or reached the maximum generation length.

---

### Why We Can Rank by Sum of Log Probabilities

In Problem Three, you used **Average Log Probability (ALP)** to compare sequences of different lengths.

During one iteration of beam search, however, all unfinished candidates have the same length because each one has just been extended by exactly one token.

Therefore, ranking candidates by:

```python
sum_log_probs
```

produces the same ordering as ranking them by ALP.

For completed sentences, which may have different lengths, continue to use **perplexity** for comparison.

---

### Beam Width and Continuation Width

Beam search uses two important parameters:

- **Beam width `B`**: the number of partial sentences retained after each iteration.
- **Continuation width `C`**: the number of possible next-token continuations generated from each partial sentence.

A larger `B` explores more possible paths but requires more computation.

A larger `C` produces more possible continuations before pruning.

In this problem, use:

```python

```

---

### Temperature Still Matters

The next-token continuations should still be sampled using **temperature-controlled sampling**, exactly as in Problems Two and Three.

Thus:

- low temperature favors high-probability continuations;
- high temperature explores a wider range of possible continuations.

However, once a token is sampled, score it using its original **untempered** log probability.

This means temperature controls **exploration**, while the beam keeps the highest-scoring partial sequences.

---

### Similar Sentences Are Expected

Beam search is designed to find high-probability sequences, not to maximize diversity.

As a result, several completed sentences may share long prefixes or differ only near the end.

This is especially likely when one partial sentence receives a strong score early in the search and several surviving beam candidates become descendants of that same prefix.

This behavior is normal and is an important contrast with ordinary temperature sampling.

````python
# ================================================
# PART A: One-Step Extension Function
# ================================================

import math
import random
import numpy as np
from heapq import heappush, heappop

def extend_partial_sentence(partial_sentence, get_next_word, temperature=1.0, random_seed=None):
    """
    Extend a partial sentence by one token using temperature-controlled sampling.
    
    Args:
        partial_sentence: (tokens_so_far, sum_log_probs) tuple
        get_next_word: The PLM dictionary
        temperature: Temperature for sampling
        random_seed: Optional seed for reproducibility
    
    Returns:
        (new_tokens, new_sum_log_probs): Extended sentence with updated score
    """
    tokens_so_far, sum_log_probs = partial_sentence
    
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    # Get left context (last 2 tokens)
    left_context = tuple(tokens_so_far[-2:])
    
    # Try trigram context first
    if left_context in get_next_word:
        log_prob_dist = get_next_word[left_context]
    else:
        # Backoff to bigram context
        left_context = tuple(tokens_so_far[-1:])
        if left_context in get_next_word:
            log_prob_dist = get_next_word[left_context]
        else:
            # Should not happen
            return None
    
    # Sample next token using temperature
    tokens = list(log_prob_dist.keys())
    log_probs = [log_prob_dist[token] for token in tokens]
    
    # Convert to ordinary probabilities
    probs = [math.exp(lp) for lp in log_probs]
    
    # Apply temperature scaling for sampling
    if temperature != 1.0:
        scaled_log_probs = [lp / temperature for lp in log_probs]
        scaled_probs = [math.exp(slp) for slp in scaled_log_probs]
        total = sum(scaled_probs)
        scaled_probs = [p / total for p in scaled_probs]
    else:
        scaled_probs = probs
    
    # Sample token
    sampled_token = rng.choices(tokens, weights=scaled_probs, k=1)[0]
    
    # Get ORIGINAL log probability
    original_log_prob = log_prob_dist[sampled_token]
    
    # Create extended sentence
    new_tokens = tokens_so_far + [sampled_token]
    new_sum_log_probs = sum_log_probs + original_log_prob
    
    return (new_tokens, new_sum_log_probs)


# ================================================
# PART B: Beam Search Implementation
# ================================================

def beam_search(get_next_word, beam_width=5, continuation_width=5, 
                temperature=1.0, max_length=40, num_completed=10, 
                random_seed=None, verbose=False):
    """
    Implement beam search to find high-probability sequences.
    
    Args:
        get_next_word: The PLM dictionary
        beam_width: Number of partial sentences to keep (B)
        continuation_width: Number of continuations per sentence (C)
        temperature: Temperature for sampling
        max_length: Maximum tokens to generate
        num_completed: Target number of completed sentences
        random_seed: Optional seed
        verbose: Whether to print debug info
    
    Returns:
        finished: List of (tokens, perplexity) for completed sentences
    """
    if random_seed is not None:
        rng = random.Random(random_seed)
    else:
        rng = random.Random()
    
    # Priority queue: store as (negative score, tokens, sum_log_probs)
    # Use negative score for max-heap behavior
    PQ = [(0.0, ['<s>'], 0.0)]
    finished = []
    
    max_iterations = max_length  # Safety limit
    
    for iteration in range(max_iterations):
        if len(finished) >= num_completed:
            break
        
        if not PQ:
            break
        
        # Generate continuations from all partial sentences in PQ
        new_candidates = []
        
        for _, tokens_so_far, sum_log_probs in PQ:
            # Generate C continuations from this partial sentence
            for cont_idx in range(continuation_width):
                # Use a different seed for each continuation
                local_seed = (random_seed + iteration * 1000 + cont_idx) if random_seed else None
                
                result = extend_partial_sentence(
                    (tokens_so_far, sum_log_probs),
                    get_next_word,
                    temperature=temperature,
                    random_seed=local_seed
                )
                
                if result is not None:
                    new_tokens, new_sum_log_probs = result
                    
                    # Check if sentence is completed
                    if new_tokens[-1] == '</s>':
                        # Compute perplexity
                        num_predictions = len(new_tokens) - 1
                        ALP = new_sum_log_probs / num_predictions
                        PP = np.exp(-ALP)
                        finished.append((new_tokens, PP))
                    else:
                        # Keep as candidate for next iteration
                        # Store as (negative score, tokens, sum_log_probs)
                        # Negative because heapq is a min-heap
                        new_candidates.append(
                            (-new_sum_log_probs, new_tokens, new_sum_log_probs)
                        )
        
        # Remove exact duplicates (optional but helpful)
        seen = set()
        unique_candidates = []
        for score, tokens, sum_log in new_candidates:
            tokens_key = tuple(tokens)
            if tokens_key not in seen:
                seen.add(tokens_key)
                unique_candidates.append((score, tokens, sum_log))
        
        # Sort by score and keep top B
        unique_candidates.sort()
        PQ = unique_candidates[:beam_width]
        
        if verbose:
            print(f"Iteration {iteration}: {len(finished)} completed, {len(PQ)} in beam")
    
    return finished


# ================================================
# PART C: Compare Three Temperatures
# ================================================

def print_sentence(tokens):
    """Reconstruct and print a human-readable sentence from tokens."""
    if not tokens:
        print("")
        return ""
    
    import string
    
    # Drop boundary markers if present
    start = 1 if tokens and tokens[0] == "<s>" else 0
    end = -1 if tokens and tokens[-1] == "</s>" else len(tokens)
    core = tokens[start:end]
    
    if not core:
        print("")
        return ""
    
    # Capitalize first token
    out = core[0].capitalize()
    
    # Process remaining tokens with spacing rules
    for i, t in enumerate(core[1:], start=1):
        is_last = (i == len(core) - 1)
        
        # Last-token punctuation rule
        if is_last and t in {'.', '?', '!'}:
            out += t
            continue
        
        # General punctuation attaches with no space
        if t in string.punctuation:
            out += t
            continue
        
        # Apostrophe-start tokens
        if t.startswith("'"):
            out += t
            continue
        
        # Capitalize special standalone words
        if t.lower() in {"i", "god"}:
            out += " " + t.capitalize()
            continue
        
        # Default: add space before normal words
        out += " " + t
    
    print(out)
    return out


print("="*70)
print("BEAM SEARCH COMPARISON")
print("="*70)

temperatures = [0.2, 1.0, 5.0]
beam_width = 5
continuation_width = 5

results_by_temp = {}

for temp in temperatures:
    print(f"\n{'='*70}")
    print(f"Temperature: {temp}")
    print(f"{'='*70}\n")
    
    # Run beam search
    finished = beam_search(
        get_next_word,
        beam_width=beam_width,
        continuation_width=continuation_width,
        temperature=temp,
        max_length=40,
        num_completed=10,
        random_seed=42,
        verbose=False
    )
    
    # Sort by perplexity
    finished.sort(key=lambda x: x[1])
    
    results_by_temp[temp] = finished
    
    print(f"Found {len(finished)} completed sentences\n")
    
    for idx, (tokens, pp) in enumerate(finished):
        print(f"Sentence {idx+1} (PP={pp:.4f}):")
        print_sentence(tokens)
        print()
    
    # Print statistics
    if finished:
        perplexities = [pp for _, pp in finished]
        print(f"Perplexity Statistics:")
        print(f"  Average: {np.mean(perplexities):.4f}")
        print(f"  Min:     {np.min(perplexities):.4f}")
        print(f"  Max:     {np.max(perplexities):.4f}")
        print()


# ================================================
# Comparative Analysis
# ================================================

print("\n" + "="*70)
print("COMPARATIVE ANALYSIS: BEAM SEARCH ACROSS TEMPERATURES")
print("="*70 + "\n")

print("""
KEY OBSERVATIONS:

1. PERPLEXITY TRENDS:
   Observe which temperature tends to produce the lowest average perplexity
   in beam search (similar to Problem 3 or different?).

2. SENTENCE SIMILARITY:
   Beam search finds high-probability sequences, so multiple sentences may
   share long common prefixes. Look for sentences that start the same way
   but diverge near the end.

3. QUALITY vs. DIVERSITY:
   Temperature 0.2 should produce high-quality (low perplexity) but similar
   sentences. Temperature 5.0 should produce more diverse but lower-quality
   sentences.

4. COMPARISON WITH TEMPERATURE SAMPLING (Problem 2):
   Beam search is more systematic and focused, while temperature sampling
   is more random. You should see higher average quality from beam search
   at each temperature.

5. BEAM STRUCTURE:
   At each iteration, beam search keeps the best 5 partial sequences.
   This means strong partial sequences tend to spawn multiple completed
   sentences that share that prefix.
""")

print("\nAverage Perplexities by Temperature (Beam Search):")
for temp in temperatures:
    if results_by_temp[temp]:
        perps = [pp for _, pp in results_by_temp[temp]]
        print(f"  Temperature {temp}: {np.mean(perps):.4f}")
````

## Appendix: Using Log Probabilities in N-Gram Models

In an N-gram language model, the probability of a sequence is obtained by multiplying the conditional probabilities of the tokens that are predicted.

For a tokenized sentence

$$
[t_1, t_2, \dots, t_n]
$$

where $t_1 = \texttt{<s>}$, the model predicts the remaining $n-1$ tokens:

$$
P(t_2, t_3, \dots, t_n \mid t_1)
=
\prod_{i=2}^{n}
P(t_i \mid \text{context}_i)
$$

For our trigram model, the context normally consists of the previous two tokens. When that trigram context is unavailable, we back off to the previous one-token bigram context.

---

### Why Use Log Probabilities?

Probabilities are numbers between 0 and 1. When many probabilities are multiplied together, the result can become extremely small.

The Austen corpus used in this homework contains approximately **432,000 tokens**, and its longest sentence contains **303 tokens**.

Suppose, just for illustration, that the average conditional probability of each predicted token were $0.1$.

For a 303-token sentence, the product would be approximately:

$$
(0.1)^{302} = 10^{-302}
$$

This is extremely small, although it is still representable using standard double-precision floating-point numbers.

But if the average conditional probability were $0.01$, then:

$$
(0.01)^{302} = 10^{-604}
$$

which is too small to represent in ordinary double precision and would **underflow to 0**.

This is why language models normally perform probability calculations in **log space**.

---

### Some Examples of Log Probabilities

We use the natural logarithm, `log`.

Since probabilities lie in the range

$$
0 \le P \le 1,
$$

their log probabilities lie in the range

$$
-\infty \le \log P \le 0.
$$

For example:

| Probability | Log Probability |
|---:|---:|
| $1.0$ | $0.0$ |
| $1/2$ | $-0.6931$ |
| $1/3$ | $-1.0986$ |
| $1/4$ | $-1.3863$ |
| $0.01$ | $-4.6052$ |
| $10^{-10}$ | $-23.0259$ |
| $10^{-1000}$ | $-2302.5851$ |

Notice that:

- more likely events have log probabilities closer to **0**;
- less likely events have increasingly negative log probabilities.

---

### The Log-Space Trick

The key identity is:

$$
\log(ab) = \log(a) + \log(b)
$$

Therefore, instead of multiplying many small probabilities:

$$
P_1 P_2 P_3 \cdots P_k,
$$

we add their log probabilities:

$$
\log P_1 + \log P_2 + \log P_3 + \cdots + \log P_k.
$$

For a generated sentence, this gives:

$$
\log P(\text{sentence})
=
\sum_{i=2}^{n}
\log P(t_i \mid \text{context}_i)
$$

In the homework, we store this quantity as:

```python
sum_log_probs
```

> Using log probabilities therefore lets us replace multiplication of very small probabilities with addition of manageable log probabilities. In this homework, `sum_log_probs` stores this accumulated score. When ordinary probabilities are needed for sampling, convert back using `exp()`. See Problem Three for how `sum_log_probs` is converted to ALP and perplexity.
